```python
def temp_cov(temp, scale):
    if scale.upper()=="C" or scale.upper()=='CELSIUS':
        print(str(temp)+" C = "+str((9/5)*float(temp)+32)+" F") 
    elif scale.upper()=='F' or scale.upper()=='FAHRENHEIT':
        print(str(temp)+ ' F = '+str((5/9)*(float(temp)-32))+' C')
    else:
        print("You made a mistake")
temp_cov(212, 'f')
```

```python
import scipy

male_w = surveys_df.where(surveys_df.sex== "M").dropna()['weight']
female_w = surveys_df.where(surveys_df.sex== "F").dropna()['weight']
scipy.stats.ttest_ind(male_w, female_w)
```

```python
male_hfl = surveys_df.where(surveys_df.sex== "M").dropna()['hindfoot_length']
female_hfl = surveys_df.where(surveys_df.sex== "F").dropna()['hindfoot_length']
scipy.stats.ttest_ind(male_hfl, female_hfl)
```
